## Modelo de Markov de segundo orden
- nueva manera de almacenar y representar las probabilidades de transicion
## implicaciones
- es que a medida que aumenta el numero de estados anteriores que consideramos, el tamaño de la matriz o tensor crece exponencialmente.
- este crecimiento podria llevar a aproblemas de eficiencia computacional y requerir recursos significativos para su implementación y calculos

```python
import pandas as pd 
import numpy as np
import string
```

```python
pa_inicial = {}
primer_orden = {}
segundo_orden = {}  
```

```python
# funcion para remover puntuacion
# s = string
def remove_punctuation(s):
    return s.translate(str.maketrans('','',string.punctuation))
```

```python
# funcion para agregar el primer orden
# d = diccionario , k = key , v = value 
def add2dict(d, k, v):
    if k not in d:
        d[k] = []
    d[k].append(v)
```

```python
# abrir archivo .txt
with open('C:/Users/ander/Escritorio/PYTHON/Python 2025/Procesamiento_lenguaje_natural_NLP/datasets/spanish_markov.txt','r', encoding='utf-8') as archivo:

    for line in archivo:
        # skip lineas vacias
        if not line.strip():
            continue

        print(line)
        # remover puntuacion
        tokens = remove_punctuation(line.rstrip().lower()).split()
        #transformando a tokens
        print(tokens)
        # tamaño de la linea
        T = len(tokens)
        print(f'tamaño de la fila: {T}')
        # recorriendo cada elemento de la fila
        for i in range(T):
            token = tokens[i]
            if i==0:
                pa_inicial[token] = pa_inicial.get(token, 0.) +1
                print(f'Palabra inicial: {token}')
            else:
                t_1 = tokens[i-1]
                if i == T-1:
                    add2dict(segundo_orden, (t_1, token), 'END')
                    print(f'Palabra final: {token}')
                if i == 1:
                    add2dict(primer_orden, t_1, token)
                    print(f'Palabra 2: {token}')  
                else:
                    t_2 = tokens[i-2]
                    add2dict(segundo_orden, (t_2, t_1), token)
                    print(f'Palabra 3: {token}')  
```

```python
# calculo de probabilidades
inicial_total = sum(pa_inicial.values())
print(inicial_total)
# t = clave , c = valor
for t, c in pa_inicial.items():
    pa_inicial[t] = c/inicial_total
print(pa_inicial)
```

```python
# funcion para diccionarios de primer y segundo orden
def list2pdict(ts):
    #diccionario vacio
    d = {}
    # tamaño de la lista
    n = len(ts)
    # ocurrencia de cada elemento de la lista
    for t in ts:
        d[t] = d.get(t, 0.) + 1
    #conversion de conteos en probabilidades relativas
    for t , c in d.items():
        d[t] = c/n
    # diccionario de probabilidades    
    return d 
```

```python
# palabras de primer orden
for t_1, ts in primer_orden.items():
    #probabilidades de primer orden
    primer_orden[t_1] = list2pdict(ts)
```

```python
# palabras de segundo orden
for k, ts in segundo_orden.items():
    segundo_orden[k] = list2pdict(ts)
```
