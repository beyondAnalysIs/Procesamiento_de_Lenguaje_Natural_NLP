### Resumen
- son esencialmente sistemas que pasan de un estado a otro con ciertas probabilidades
- estas probabilidades no dependen de como llego al estado actual, sino solo del estado actual en sí

### Representación matamática
- si tenemos tres estados, A,B y C, la probabilidad de pasar de A a B podria ser representada por P(B|A)
- A representa qu eesta soleado y B que esta nublado, P(B|A) es la probabilidad de que se ve vuelva nublado dado que ghoy esta soleado

### Uso de los modelos de Markov para secuencias
- para predecur la siguiente palabra basandote en la actual
- ej. después de la palabra 'buenos', la palabra 'dias' aparece un 80% de las veces, entonces: P('dias'|'buenos') = 0.8
